```python
def calc_stock(delivered: int = 0, sold: int = 0, returned: int = 0):
    if not (isinstance(delivered, (int, float)) and
            isinstance(sold, (int, float)) and
            isinstance(returned, (int, float))):
        print("Eingabefehler, nur Zahlen erlaubt!")
        return

    bestand = 500
    final_stock = bestand + delivered + returned - sold

    print(f"Nach der Lieferung von {delivered}, dem Verkauf von {sold} und {returned} Retouren beträgt der Bestand {final_stock}.")

calc_stock("st", 25, 5)
calc_stock(20, 25, 5)


```

```python
def filter_weights(data: list[int]) -> list[int]:
    selected_weights: list[int] = []

    for e in data:
        if e % 2 == 0 and e % 3 == 0 and 25 <= e <= 36:
            selected_weights.append(e)

    print(f"Die Liste selected_weights beinhaltet folgende Einträge: {selected_weights}.")

    for e in selected_weights:
        if e % 5 == 0:
            selected_weights.remove(e)

    print(f"Finale Liste nach Bereinigung: {selected_weights}.")

    return selected_weights

ergebnis = filter_weights([12,18,26,27,32,30,36,42,24,60,48])
```
